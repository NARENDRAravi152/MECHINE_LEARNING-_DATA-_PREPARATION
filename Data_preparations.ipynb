## METHOD -2 ONE-HOT-ENCODER

```python
weather_data_copy_3=weather_data.copy()
weather_data_copy_3
```

```python
ohe_encoder=OneHotEncoder()
ohe_encoder
```

```python
ohe_encoder.fit_transform(X=weather_data_copy_3,y=weather_data_copy_3[["Weather"]])
```

```python
ohe_encoder.fit_transform(X=weather_data_copy_3,y=weather_data_copy_3["Weather"].values)
```

```python
ohe_encoder.fit_transform(X=weather_data_copy_3,y=weather_data_copy_3["Weather"].to_numpy)
```

```python
weather_encoder=ohe_encoder.fit_transform(weather_data_copy_3[["Weather"]])
weather_encoder
```

```python
weather_encoder.toarray()
```

```python
ohe_encoder.get_feature_names_out(["Weather"])
```

```python
weather_encoders=pd.DataFrame(data=weather_encoder.toarray(),columns=ohe_encoder.get_feature_names_out(["Weather"]))
weather_encoders
```

```python
weather_data_copy_3=weather_data_copy_3.reset_index(drop=True)
weather_encoders=weather_encoders.reset_index(drop=True)
weather_data_copy_3_final=pd.concat([weather_data_copy_3,weather_encoders],axis=1)
```
